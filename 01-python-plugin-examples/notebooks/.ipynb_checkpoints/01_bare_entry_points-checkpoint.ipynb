````python
from importlib.metadata import entry_points

data = {"name": "Alice", "role": "Engineer", "level": 4}
````

## Discover all formatters

````python
eps = entry_points(group="myapp.formatters")
for ep in eps:
    print(f"{ep.name:8} → {ep.value}")
````

## Load and use them

````python
formatters = {}
for ep in entry_points(group="myapp.formatters"):
    try:
        formatters[ep.name] = ep.load()
    except Exception as e:
        print(f"Failed to load {ep.name}: {e}")

print("Loaded:", list(formatters))
````

````python
for name, cls in sorted(formatters.items()):
    print(f"\n=== {name} ===")
    print(cls().format(data))
````
