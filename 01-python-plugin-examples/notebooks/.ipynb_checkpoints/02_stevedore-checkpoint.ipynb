````python
from stevedore import driver, extension

data = {"name": "Alice", "role": "Engineer", "level": 4}
````

## Load a single driver

````python
mgr = driver.DriverManager(
    namespace="myapp.formatters",
    name="json",
    invoke_on_load=True,
)
print(mgr.driver.format(data))
````

## Load all extensions and call a method on each

````python
mgr = extension.ExtensionManager(
    namespace="myapp.formatters",
    invoke_on_load=True,
)

results = mgr.map(lambda ext, d: (ext.name, ext.obj.format(d)), data)

for name, output in results:
    print(f"\n=== {name} ===")
    print(output)
````

## List available names

````python
print(mgr.names())
````
