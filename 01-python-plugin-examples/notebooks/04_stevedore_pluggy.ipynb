# Stevedore + Pluggy

This example combines Stevedore for plugin discovery and construction with Pluggy for hook specification and invocation.

## Install the example

From the `04-stevedore-pluggy` directory, install the package in editable mode:

```bash
pip install -e .
```

````python
import sys
sys.path.insert(0, "../04-stevedore-pluggy")

import pluggy
from stevedore import extension

from myapp.hooks import FormatterSpec

data = {"name": "Alice", "role": "Engineer", "level": 4}
````

## Discover plugins with Stevedore

Stevedore reads the `myapp.formatters` entry-point group and constructs each plugin.

````python
manager = extension.ExtensionManager(
    namespace="myapp.formatters",
    invoke_on_load=True,
)

for ext in manager:
    print(ext.name, type(ext.obj).__name__)
````

## Register the plugins with Pluggy

The objects created by Stevedore become Pluggy plugin implementations. Pluggy provides the behavioral contract and hook invocation.

````python
pm = pluggy.PluginManager("myapp")
pm.add_hookspecs(FormatterSpec)

for ext in manager:
    pm.register(ext.obj, name=ext.name)

print(pm.list_name_plugin())
````

## Call the `format_data` hook

````python
results = pm.hook.format_data(data=data)

for result in results:
    print(result)
    print("---")
````

## Inspect the hook

Pluggy can also expose the hook specification and registered implementations.

````python
hook = pm.hook.format_data

print("Hook:", hook)
print("Implementations:", hook.get_hookimpls())
````
