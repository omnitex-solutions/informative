````python
import pluggy
import sys
sys.path.insert(0, "../03-pluggy")

from myapp.hooks import FormatterSpec
from plugins import plain, json_fmt, table

data = {"name": "Alice", "role": "Engineer", "level": 4}
````

## Create the PluginManager and register plugins

````python
pm = pluggy.PluginManager("myapp")
pm.add_hookspecs(FormatterSpec)

pm.register(plain)
pm.register(json_fmt)
pm.register(table)

print("Registered plugins:", pm.get_plugins())
````

## Call the format_data hook on all implementations

````python
results = pm.hook.format_data(data=data)

for result in results:
    print(result)
    print("---")
````
